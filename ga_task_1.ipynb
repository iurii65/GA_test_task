##### Привязка наценки и расчет цены

NB! Предоставленное решение задачи по проставлению наценки выполняется за время O(n*k), при n == k время решения будет O(n^2). В настощее время я изучаю алгоритмы, поэтому вопрос о решении данной задачи за время O(n) возник сразу. Погуглив и пообщавшись с gpt выяснил, что эту задачу можно попробовать решить с помощью алгоритма DFS поиска в глубину. Увы до данной темы в обучении я ещё не дошёл, поэтому не стал использовать ИИшный код без его понимания и возможности изменить и прокомментировать. 

```python
def find_markup(row, markup):
    # Продукт
    if not markup[markup['Продукт'] == row['Код товара']].empty:
        return markup[markup['Продукт'] == row['Код товара']].iloc[0]['Наценка'], 1

    # 2 Бренд/Категория3
    if not markup[(markup['Бренд'] == row['Бренд']) & (markup['Категория'] == row['Категория3']) & (markup['Сегмент рынка'].isna()) & (markup['Продукт'].isna())].empty:
        return markup[(markup['Бренд'] == row['Бренд']) & (markup['Категория'] == row['Категория3']) & (markup['Сегмент рынка'].isna()) & (markup['Продукт'].isna())].iloc[0]['Наценка'], 2

    # 3 Бренд/Категория2
    if not markup[(markup['Бренд'] == row['Бренд']) & (markup['Категория'] == row['Категория2']) & (markup['Сегмент рынка'].isna()) & (markup['Продукт'].isna())].empty:
        return markup[(markup['Бренд'] == row['Бренд']) & (markup['Категория'] == row['Категория2']) & (markup['Сегмент рынка'].isna()) & (markup['Продукт'].isna())].iloc[0]['Наценка'], 3

    # 4 Бренд/Категория1
    if not markup[(markup['Бренд'] == row['Бренд']) & (markup['Категория'] == row['Категория1']) & (markup['Сегмент рынка'].isna()) & (markup['Продукт'].isna())].empty:
        return markup[(markup['Бренд'] == row['Бренд']) & (markup['Категория'] == row['Категория1']) & (markup['Сегмент рынка'].isna()) & (markup['Продукт'].isna())].iloc[0]['Наценка'], 4

    # 5 Бренд
    if not markup[(markup['Бренд'] == row['Бренд']) & (markup['Категория'].isna()) & (markup['Сегмент рынка'].isna()) & (markup['Продукт'].isna())].empty:
        return markup[(markup['Бренд'] == row['Бренд']) & (markup['Категория'].isna()) & (markup['Сегмент рынка'].isna()) & (markup['Продукт'].isna())].iloc[0]['Наценка'], 5

    # 6
    if not markup[(markup['Сегмент рынка'] == row['Сегмент']) & (markup['Категория'] == row['Категория3']) & (markup['Бренд'].isna()) & (markup['Продукт'].isna())].empty:
        return markup[(markup['Сегмент рынка'] == row['Сегмент']) & (markup['Категория'] == row['Категория3']) & (markup['Бренд'].isna()) & (markup['Продукт'].isna())].iloc[0]['Наценка'], 6

    # 7
    if not markup[(markup['Сегмент рынка'] == row['Сегмент']) & (markup['Категория'] == row['Категория2']) & (markup['Бренд'].isna()) & (markup['Продукт'].isna())].empty:
        return markup[(markup['Сегмент рынка'] == row['Сегмент']) & (markup['Категория'] == row['Категория2']) & (markup['Бренд'].isna()) & (markup['Продукт'].isna())].iloc[0]['Наценка'], 7

    # 8
    if not markup[(markup['Сегмент рынка'] == row['Сегмент']) & (markup['Категория'] == row['Категория1']) & (markup['Бренд'].isna()) & (markup['Продукт'].isna())].empty:
        return markup[(markup['Сегмент рынка'] == row['Сегмент']) & (markup['Категория'] == row['Категория1']) & (markup['Бренд'].isna()) & (markup['Продукт'].isna())].iloc[0]['Наценка'], 8

    # 9
    if not markup[(markup['Сегмент рынка'] == row['Сегмент']) & (markup['Категория'].isna()) & (markup['Бренд'].isna()) & (markup['Продукт'].isna())].empty:
        return markup[(markup['Сегмент рынка'] == row['Сегмент']) & (markup['Категория'].isna()) & (markup['Бренд'].isna()) & (markup['Продукт'].isna())].iloc[0]['Наценка'], 9

    # nan если ничего выше
    return np.nan, 99
```

```python
# запуск функции выше
results = list_product.apply(lambda row: find_markup(row, markup), axis=1)
```

##### Сравнительный анализ цен

Есть два основных подхода:<br>1. Для каждой номенклатуры присвоить собственный уровень наценки, данный подход позволит гибко придерживаться стратегии компании, но в тоже время не занижать цены по товарам которые и так ниже чем у конкурентов.<br>2. Наценка регулируется для группы товаров, при такой схеме возможна ситуация, что из-за товара с высокой себестоимостью придется сильно снижать маржинальность остальных товаров в этой группе. Для тестовой задачи разделим популяцию, выберем две группы для предложения внесения изменения наценки в справочник. По остальным товаром возьмем минимальную цену конкурентов и проверим, что наценка не ниже 10%.

```python
#возьмем группу где наценка на весь бренд (5)

analysis_5 = analysis_preces[(analysis_preces['cur_priority'] == 5) & (analysis_preces['Бренд'] == 'Clarins')].copy()

#текущая наценка на бренд 103%
analysis_5['new_markup'] = 0.1
analysis_5['new_price'] = analysis_5['Цена закупки товара'] * (analysis_5['new_markup'] + 1)


analysis_5_false = analysis_5[analysis_5['new_price'] > analysis_5['Мин_цена_конкурента']]

analysis_5_false.head(2)
```

NB! Как было сказано выше из-за того что группа на которую действует одна наценка очень большая и разнородная. Невозможно применить одну наценку на все товары чтобы удовлетворить условия (цена ниже конкурентов и наценка выше 10%) 

```python
#наценка на бренд и категория 3

analysis_2 = analysis_preces[(analysis_preces['cur_priority'] == 2) &
                            (analysis_preces['Бренд'] == 'Clarins') &
                            (analysis_preces['Категория3'] == 'Крем для тела')
].copy()

#текущая наценка на бренд и категорию 3 106%
analysis_2['new_markup'] = 0.35
analysis_2['new_price'] = analysis_2['Цена закупки товара'] * (analysis_2['new_markup'] + 1)


analysis_2_false = analysis_2[analysis_2['new_price'] > analysis_2['Мин_цена_конкурента']]

analysis_2_false.head(2)
```

Для того чтобы следовать стретегии компании необходимо в справочник наценок внести новую наценку по бренду Clarins и категории3 'Крем для тела'.
Однако при таком способе регулировании мы теряем маржинальность у товаров чья цена первонаально уже была ниже чем у конкуретов.

```python
#Для всех остальных товаров сделаем индивидуальную наценку

analysis_other = analysis_preces[(~analysis_preces['Код товара'].isin(analysis_2['Код товара'])) &
                                    (analysis_preces['diff'] > 100)].copy()


analysis_other['new_price'] = analysis_other['Мин_цена_конкурента']
analysis_other['new_markup'] = (analysis_other['new_price'] - analysis_other['Цена закупки товара'])/analysis_other['Цена закупки товара']
```

```python
# товары по которым невозможно выполнить условия самой низкой цены в рынке и наценки не менее 10% 

analysis_error = analysis_other[analysis_other['new_markup'] < 0.1]
print(f'Кол-во товаров с невозможностью выполнения сразу двух условий: {len(analysis_error)}')
analysis_error.head(2)
```
